**Introduction**
- This project investigates how college students’ digital professional identities, as presented on GitHub, relate to their academic majors and class years. Our theoretical model posits that a student field of study (major) and academic progression (cohort) are significant factors that influence their self-presentation on professional platforms. The central theoretical question examines if and how these variables affect how many social media platforms are linked on a GitHub profile. Our findings suggest that a person’s year and major do influence their Github presentation, with certain majors more consistently linking to platforms like LinkedIn.our data only reflects declared links, not actual social media use, which may lead to an underrepresentation of activity, which is a key limitation in our study. Future research would benefit from longitudinal data across multiple semesters to better track evolving trends and a clearer breakdown of academic concentrations.  


**Theory**
- Our theoretical model is finding out if there’s correlation between the age of a student and how many socials they have present in their github profile. It tests the amount of professionalism a student has in their profile as well as the amount of networking made available due to their personal contacts being present on their github. Our question asks if there’s a correlation between those who are closer to getting a career expressing their socials as to those who are younger and not quite looking for a career. Our theoretical question is relevant because it shows if older students are more likely to put their social accounts up compared to those who are younger. It also compares the professionalism between older students and younger ones. 



**Data Model**
- Attribute name: cohort
- Attribute data type: string type with a controlled vocabulary and variables for majors and gender including boolean style for TRUE and FALSE [freshman, sophomore, junior, senior, not indicated]
- Attribute description: This variable collects what cohort the student is part of. It is only listed if it is explicitly indicated in the GitHub profile.




**Analysis**
- We are cross-tabulating the variables “major” and “social media links” to reveal patterns in digital self presentation across different majors. We are hoping to show whether students in certain fields of study are more or less likely to curate their GitHub profiles with links to both professional and social platforms like LinkedIn or Instagram. We hypothesize that majors with a stronger direct link to the tech industry or digital communication (like Computer Science or Marketing) will demonstrate a higher frequency of including these links, using GitHub as a part of their professional portfolio.
